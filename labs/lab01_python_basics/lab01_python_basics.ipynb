## 2) Data Types

Python has many built-in data types. We will use `type()` to check them.


````python
# Numbers: int (integer) and float (decimal)
a = 7        # int
b = 2.5      # float
c = 3
print(a, "is", type(a))
print(b, "is", type(b))
````

````python
# Data types while performing mathematical operations- it depends on the operation and the operand types. 
x = a+b
print (x, "is", type(x))
````

````python
y = a+c
print (y, "is", type(y))
````

````python
z = a/c
print (z, "is", type(z))
# Division (/) → always float
# Mixing int and float → result is float
````

````python
# Basic arithmetic
print("a + 3 =", a + 3)
print("a / 2 =", a / 2)     # division produces a float
print("a % 2 =", a % 2)     # remainder (modulo)
print("a ** 2 =", a ** 2)   # exponent
````

### Strings

Strings are sequences of characters. They can be created using single quotes `'...'` or double quotes `"..."`.


````python
# Strings
name = "SAGE"  # change this to your name
course = 'Intro to Python'

print("name is", type(name))
print("course is", type(course))

# String concatenation (joining)
print("Hello, " + name + "!")

# f-strings (recommended): easier formatting
print(f"Welcome, {name}, to {course}.")

# Indexing (strings are sequences)
text = "Python"
print("text =", text)
print("First character:", text[0])
print("Last character:", text[-1])

````

````python
# Slicing
print("First 3 characters:", text[0:3])
print("Everything except first:", text[1:])

# Useful string methods
sentence = "  Python is fun!  "
print("Original:", repr(sentence))
print("strip():", repr(sentence.strip()))
print("upper():", sentence.upper())
print("replace():", sentence.replace("fun", "powerful"))
````

### Printing

`print()` displays text or values. You can print multiple things separated by commas.


````python
x = 5
y = 12

# Printing multiple values
print("x =", x, "| y =", y)

# Formatting with f-strings (cleaner)
print(f"x = {x} | y = {y} | x+y = {x+y}")
````

### Lists

Lists store ordered collections of items and are **mutable** (you can change them).


````python
# Lists
nums = [10, 20, 30]
mixed = [1, "two", 3.0, True]

print("nums:", nums)
print("mixed:", mixed)

# Indexing
print("First element of nums:", nums[0])
print("Last element of nums:", nums[-1])

# Modify an element
nums[1] = 99
print("After modification:", nums)

# Append adds to the end
nums.append(40)
print("After append:", nums)

# Length of a list
print("Length of nums:", len(nums))

# Slicing
print("First two elements:", nums[:2])

# Sorting (in-place)
nums_sorted = nums.copy()
nums_sorted.sort()
print("Sorted copy:", nums_sorted)
print("Original nums still:", nums)
````

### Dictionaries

Dictionaries store **key → value** pairs. Keys are usually strings or numbers.


````python
# Dictionaries
student = {
    "name": "Aisha",
    "major": "Geography",
    "year": 2
}

print("student:", student)
print("Keys:", list(student.keys()))
print("Values:", list(student.values()))

# Access a value by key
print("Name:", student["name"])

# Add or update a key
student["year"] = 3
student["gpa"] = 3.7

print("Updated student:", student)

# Safe access with .get() (avoids KeyError)
print("Minor (maybe missing):", student.get("minor", "Not provided"))
````

### Booleans

Booleans represent truth values: `True` or `False`.
They are commonly produced by comparisons.


````python
is_sunny = True
is_raining = False

print(is_sunny, type(is_sunny))
print(is_raining, type(is_raining))

# Boolean logic
print("sunny AND raining:", is_sunny and is_raining)
print("sunny OR raining:", is_sunny or is_raining)
print("NOT sunny:", not is_sunny)
````

### Tuples

Tuples are ordered collections like lists, but **immutable** (cannot be changed).
They are often used for fixed-size records. does not support item assignment


````python
# Tuples
point = (3, 7)

print("point:", point)
print("x-coordinate:", point[0])
print("y-coordinate:", point[1])

# Uncommenting the next line would cause an error (tuples are immutable):
# point[0] = 10
````

### Sets

Sets store **unique** items (no duplicates) and are unordered.
Useful for membership testing and removing duplicates.


````python
# Sets
items = [1, 2, 2, 3, 3, 3, 4]
unique_items = set(items)

print("Original list:", items)
print("Unique set:", unique_items)

# Membership testing
print("Is 3 in the set?", 3 in unique_items)
print("Is 10 in the set?", 10 in unique_items)

# Set operations
A = {1, 2, 3}
B = {3, 4, 5}

print("A union B:", A | B)
print("A intersection B:", A & B)
print("A difference B:", A - B)
````

## 3) Comparison Operators

Comparisons return booleans (`True`/`False`).
Common operators:

- `==` equal to  
- `!=` not equal to  
- `<`, `<=`, `>`, `>=`  
- `in` membership (for lists/strings/sets/dicts)



````python
x = 10
y = 12

print("x == y:", x == y)
print("x != y:", x != y)
print("x < y:", x < y)
print("x >= 10:", x >= 10)

# Membership
letters = ["a", "b", "c"]
print("'b' in letters:", "b" in letters)
print("'z' in letters:", "z" in letters)

# For dictionaries, 'in' checks keys
student = {"name": "Aisha", "year": 2}
print("'name' in student:", "name" in student)
print("'gpa' in student:", "gpa" in student)
````

## 4) `if`, `elif`, `else` Statements

Use conditionals to make decisions.
Python uses indentation (spaces) to define blocks.


````python
# Example: grading logic
score = 82  # try changing this number

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "D or F"

print(f"Score = {score} → Grade = {grade}")
````

## 5) `for` Loops

A `for` loop repeats over items in a sequence (like a list) one at a time.


````python
# Loop over a list
fruits = ["apple", "banana", "mango"]

for fruit in fruits:
    print("Fruit:", fruit)


# Loop over a string (character by character)
word = "data"
for ch in word:
    print("Character:", ch)

seq = [1,2,3,4,5]
for item in seq:
    print(item)
````

## 6) `while` Loops

A `while` loop repeats **as long as** a condition is True.
Be careful: if the condition never becomes False, the loop will run forever.


````python
# Count from 1 to 5 using a while loop
count = 1

while count <= 5:
    print("count =", count)
    count += 1  # same as count = count + 1
````

## 7) `range()`

`range()` generates a sequence of integers, commonly used with `for` loops.

Examples:
- `range(5)` → 0,1,2,3,4  
- `range(2, 6)` → 2,3,4,5  
- `range(0, 10, 2)` → 0,2,4,6,8 (step of 2)


````python
# range examples
print("range(5):", list(range(5)))
print("range(2, 6):", list(range(2, 6)))
print("range(0, 10, 2):", list(range(0, 10, 2)))

# Use range in a loop
total = 0
for i in range(1, 6):  # 1 through 5
    total += i

print("Sum of 1..5 =", total)
````

## 8) Functions

Functions package reusable logic.

### Why functions?
- Avoid repeating code
- Make code easier to read and test

Syntax:
```python
def function_name(parameters):
    # code
    return result
```


````python
# A simple function
def add_two_numbers(a, b):
    """Return the sum of a and b."""
    return a + b

print(add_two_numbers(3, 4))
print(add_two_numbers(10, -2))
````

````python
# Function with a default argument
def greet(name, course="Python Basics"):
    """Print a greeting message."""
    print(f"Hello, {name}! Welcome to {course}.")

greet("Aisha")
greet("Jennifer", course="Data Science 101")
````
